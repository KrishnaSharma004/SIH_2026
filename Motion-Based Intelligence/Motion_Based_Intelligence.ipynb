```python
import sys
sys.path.append('/content/ByteTrack')
from yolox.tracker.byte_tracker import BYTETracker
import numpy as np

class TrackerArgs:
    track_thresh = 0.5
    track_buffer = 30
    match_thresh = 0.8
    mot20 = False

tracker = BYTETracker(TrackerArgs())
```

```python
import numpy as np
if not hasattr(np, 'float'):
    np.float = float
if not hasattr(np, 'int'):
    np.int = int
if not hasattr(np, 'bool'):
    np.bool = bool
if not hasattr(np, 'object'):
    np.object = object
```

```python
cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter('tracked_output.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))

trajectories = {}  # track_id -> list of (frame_idx, cx, cy, timestamp)
frame_idx = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    yolo_result = model.predict(frame, classes=[2,3,5,7], verbose=False)[0]
    boxes = yolo_result.boxes.xyxy.cpu().numpy()
    scores = yolo_result.boxes.conf.cpu().numpy()

    if len(boxes) > 0:
        dets = np.hstack((boxes, scores.reshape(-1, 1)))
    else:
        dets = np.empty((0, 5))

    online_targets = tracker.update(dets, [h, w], [h, w])

    timestamp = frame_idx / fps
    for t in online_targets:
        tid = t.track_id
        x, y, bw, bh = t.tlwh
        cx, cy = x + bw / 2, y + bh / 2

        if tid not in trajectories:
            trajectories[tid] = []
        trajectories[tid].append((frame_idx, cx, cy, timestamp))

        cv2.rectangle(frame, (int(x), int(y)), (int(x+bw), int(y+bh)), (0, 255, 0), 2)
        cv2.putText(frame, f"ID:{tid}", (int(x), int(y)-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    out.write(frame)
    frame_idx += 1

cap.release()
out.release()
print(f"Done. Tracked {len(trajectories)} unique vehicles.")
```

```python
import pandas as pd

def compute_motion_features(traj, fps, px_to_meter=0.05):
    """traj: list of (frame_idx, cx, cy, timestamp)"""
    df = pd.DataFrame(traj, columns=["frame", "cx", "cy", "t"])
    df["dx"] = df["cx"].diff()
    df["dy"] = df["cy"].diff()
    df["dt"] = df["t"].diff()
    df["dist_px"] = np.sqrt(df["dx"]**2 + df["dy"]**2)
    df["speed_px_s"] = df["dist_px"] / df["dt"]
    df["speed_m_s"] = df["speed_px_s"] * px_to_meter  # rough calibration
    df["accel"] = df["speed_m_s"].diff() / df["dt"]
    df["heading"] = np.arctan2(df["dy"], df["dx"])
    df["heading_change"] = df["heading"].diff().abs()
    return df

# Example: analyze one tracked vehicle
sample_id = list(trajectories.keys())[0]
motion_df = compute_motion_features(trajectories[sample_id], fps)
motion_df.head(10)
```

```python
def flag_rash_driving(motion_df, speed_thresh=15, accel_thresh=8, heading_thresh=0.5):
    flags = []
    for i, row in motion_df.iterrows():
        is_rash = False
        reasons = []
        if row["speed_m_s"] > speed_thresh:
            is_rash = True
            reasons.append("overspeed")
        if abs(row["accel"]) > accel_thresh:
            is_rash = True
            reasons.append("harsh_accel/brake")
        if row["heading_change"] > heading_thresh:
            is_rash = True
            reasons.append("swerving")
        flags.append({"frame": row["frame"], "flagged": is_rash, "reasons": reasons})
    return pd.DataFrame(flags)

rash_flags = flag_rash_driving(motion_df)
rash_flags[rash_flags["flagged"] == True]
```

```python
all_events = []
for tid, traj in trajectories.items():
    if len(traj) < 5:  # skip very short tracks
        continue
    m_df = compute_motion_features(traj, fps)
    r_df = flag_rash_driving(m_df)
    flagged = r_df[r_df["flagged"] == True]
    if not flagged.empty:
        all_events.append({"track_id": tid, "num_flagged_frames": len(flagged),
                            "reasons": flagged["reasons"].tolist()})

pd.DataFrame(all_events)
```

```python
def detect_incident_candidates(motion_df, track_id, accel_thresh=10):
    incidents = []
    for i, row in motion_df.iterrows():
        if pd.notna(row["accel"]) and abs(row["accel"]) > accel_thresh:
            incidents.append({
                "track_id": track_id,
                "frame": row["frame"],
                "timestamp": row["t"],
                "type": "harsh_decel_possible_collision"
            })
    return incidents

incidents = detect_incident_candidates(motion_df, sample_id)
incidents
```

```python
import json

def build_event_json(track_id, frame, timestamp, event_type, confidence, gps=None):
    return {
        "track_id": int(track_id),
        "event_type": event_type,
        "frame": int(frame),
        "timestamp_sec": float(timestamp),
        "gps_location": gps if gps else {"lat": None, "lon": None},
        "confidence": confidence
    }

events_json = []
for e in all_events:
    events_json.append(build_event_json(e["track_id"], 0, 0, "rash_driving", 0.8))

with open("motion_events.json", "w") as f:
    json.dump(events_json, f, indent=2)

files.download("motion_events.json")
```
